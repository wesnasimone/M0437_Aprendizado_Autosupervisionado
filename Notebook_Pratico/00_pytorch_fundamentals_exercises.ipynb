### 2. Create a random tensor with shape `(7, 7)`.


````python
# Import torch
import torch

# Create random tensor
M_random = torch.rand(size=(7,7))
M_random, M_random.size()
````

### 3. Perform a matrix multiplication on the tensor from 2 with another random tensor with shape `(1, 7)` (hint: you may have to transpose the second tensor).

````python
# Create another random tensor
M_random2 = torch.rand(size=(1,7))

# Perform matrix multiplication
result = torch.matmul(M_random, M_random2.T)

print(f"Matriz A: {M_random.shape}")
print(f"Matriz B: {M_random2.shape}")
print(f"Matriz B transposta: {M_random2.T.shape}")
print(f"Dimensão da Resposta: {result.shape}")
result
````

### 4. Set the random seed to `0` and do 2 & 3 over again.

The output should be:
```
(tensor([[1.8542],

         [2.2884],
         [3.0481],
         [1.7067],
         [2.5290],
         [1.7989]]), torch.Size([7, 1]))
```


### 6. Create two random tensors of shape `(2, 3)` and send them both to the GPU (you'll need access to a GPU for this). Set `torch.manual_seed(1234)` when creating the tensors (this doesn't have to be the GPU random seed). The output should be something like:


### 10. Make a random tensor with shape `(1, 1, 1, 10)` and then create a new tensor with all the `1` dimensions removed to be left with a tensor of shape `(10)`. Set the seed to `7` when you create it and print out the first tensor and it's shape as well as the second tensor and it's shape.

The output should look like:

```
tensor([[[[0.5349, 0.1988, 0.6592, 0.6569, 0.2328, 0.4251, 0.2071, 0.6297,
           0.3653, 0.8513]]]]) torch.Size([1, 1, 1, 10])
tensor([0.5349, 0.1988, 0.6592, 0.6569, 0.2328, 0.4251, 0.2071, 0.6297, 0.3653,
        0.8513]) torch.Size([10])
```

````python
# Set seed
torch.manual_seed(7)

# Create random tensor
Tensor = torch.rand(1, 1, 1, 10)

# Remove single dimensions
Tensor_new = Tensor.squeeze()

# Print out tensors and their shapes
print(Tensor)
print(Tensor.size())
print(Tensor_new)
print(Tensor_new.size())
````
